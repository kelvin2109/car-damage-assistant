```python
img_list = []
def load_img(path):
    img_path = "data/train/"
    image = cv2.imread(os.path.join(img_path,path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    return image
```

```python
#don't need to add normalization since its already handle by detectron2
transform = A.Compose(
    [
        A.Affine(
            scale=(0.8, 1.2),
            translate_percent=(-0.1,0.1),
            rotate=(-30, 30),
            shear=(-10,10),
            p=0.5
        ),
        A.OneOf([
            A.RandomRain( p=0.5),
            A.RandomSnow(p=0.5)
        ],p=0.4),

        A.OneOf([
            A.RandomBrightnessContrast(
                brightness_limit=0.1,
                contrast_limit=0.1
            ),
            A.RandomGamma(
                gamma_limit=(90, 110)
            )
        ], p=0.2)
        
    ],    bbox_params=A.BboxParams(format='coco', label_fields=['labels']))
```
